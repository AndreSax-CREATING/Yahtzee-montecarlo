```python
import random

def tirar_los_dados(num_dados=5):
  """Porfavor simula el lanzamiento de 5 dados de 6 caras"""
  return [random.randint(1,6) for _ in range(num_dados)]

def calcular_puntaje_simple(dados):
  """Calcula un puntaje basico de suma total para simplificar la simulacion"""
  return sum(dados)

def simular_partida_yahtzee():
  """Simulamos una ronda de Yahtzee para dos jugadores aplicando Montecarlo."""
  # Jugador 1
  dados_j1 = tirar_los_dados()
  #Simulacion de 3 reintentos aleatorios de algunos dados

  for _ in range(2):
    dados_j1 = [r if random.random() > 0.5 else random.randint(1, 6) for r in dados_j1  ]

  # Jugador 2
  dados_j2 = tirar_los_dados()
  for _ in range(2):
    dados_j2 = [r if random.random() > 0.5 else random.randint(1,6) for r in dados_j2 ]

    return calcular_puntaje_simple(dados_j1), calcular_puntaje_simple(dados_j2)

def ejecutar_montecarlo(iteraciones=10000):
    """Ejecuta el método de Montecarlo con miles de iteraciones."""
    victorias_j1 = 0
    victorias_j2 = 0
    empates = 0
    suma_total_puntos = 0

    for _ in range(iteraciones):
        p1, p2 = simular_partida_yahtzee()
        suma_total_puntos += (p1 + p2)
        if p1 > p2:
            victorias_j1 += 1
        elif p2 > p1:
            victorias_j2 += 1
        else:
            empates += 1

    print(f"--- RESULTADOS DE LA SIMULACIÓN (Montecarlo: {iteraciones} iteraciones) ---")
    print(f"Victorias Jugador 1: {victorias_j1} ({victorias_j1/iteraciones*100:.2f}%)")
    print(f"Victorias Jugador 2: {victorias_j2} ({victorias_j2/iteraciones*100:.2f}%)")
    print(f"Empates: {empates} ({empates/iteraciones*100:.2f}%)")
    print(f"Promedio de puntos por tirada (Ley de los Grandes Números): {suma_total_puntos/(iteraciones*2):.2f}")

# Ejecutamos la simulación con 10,000 partidas
ejecutar_montecarlo(10000)



```
